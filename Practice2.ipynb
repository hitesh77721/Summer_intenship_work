```python
text = "Hello!!! 😊😊 My name is Hitesh. I'm learning NLP in 2026. Visit https://google.com"
original_text = text.copy()
```

```python
# lowering
text = text.lower()
```

```python
# removing url

import re 
text = re.sub(r'https?://\S+|www\.\S+', '', text)
text
```

```python
import emoji
text = emoji.replace_emoji(text, replace='')
text
```

```python
# expand contraction 

import contractions
text = contractions.fix(text)
text
```

```python
# removing punctuation 

import string
# string.punctuation
table = str.maketrans('', '', string.punctuation)
text = text.translate(table)
text
```

```python
# remove numbers 
# text = re.sub(r'\d+', '', text)
# text
```

```python
from nltk.tokenize import word_tokenize

import nltk
nltk.download('punkt')

tokens = word_tokenize(text)
```

```python
tokens
# text
```

```python
import nltk

nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
stop_words
```

```python
filtered_tokens = []

for word in tokens:
    if word not in stop_words:
        filtered_tokens.append(word)
```

```python
filtered_tokens
```

```python
from nltk.stem import PorterStemmer 
from nltk.stem import WordNetLemmatizer

ps = PorterStemmer()
lemmatizer = WordNetLemmatizer()

filtered_tokens += ["studying"]
stemmed_words = []
lemmatizer_words = []
for word in filtered_tokens:
    stemmed_words.append(ps.stem(word))
    lemmatizer_words.append(lemmatizer.lemmatize(word, pos="v"))
```

```python
print(lemmatizer_words)
filtered_tokens, stemmed_words
```
